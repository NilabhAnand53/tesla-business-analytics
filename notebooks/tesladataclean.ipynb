```python
# Calculate the expected total order value from orders_df
orders_df['total_order_value'] = orders_df['quantity'] * orders_df['unit_price_usd']

# Merge this expected value into revenue_df on order_id
revenue_df = revenue_df.merge(
    orders_df[['order_id', 'total_order_value']],
    on='order_id',
    how='left'
)

# Fill missing amount_paid_usd with total_order_value
missing_before = revenue_df['amount_paid_usd'].isna().sum()
revenue_df['amount_paid_usd'] = revenue_df['amount_paid_usd'].fillna(revenue_df['total_order_value'])
missing_after = revenue_df['amount_paid_usd'].isna().sum()

print(f"Missing amount_paid_usd values before: {missing_before}")
print(f"Missing amount_paid_usd values after filling from order values: {missing_after}")

# Drop the helper total_order_value column if desired
revenue_df = revenue_df.drop(columns=['total_order_value'])

display(revenue_df.head(10))
```
